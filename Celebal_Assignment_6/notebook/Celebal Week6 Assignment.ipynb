# STEP 5: Create Manual Schema

```python
myschema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("employee_name", StringType(), True),
    StructField("gender", StringType(), True),
    StructField("age", StringType(), True),
    StructField("department", StringType(), True),
    StructField("designation", StringType(), True),
    StructField("salary", StringType(), True),
    StructField("experience", StringType(), True),
    StructField("joining_date", StringType(), True),
    StructField("city", StringType(), True),
    StructField("state", StringType(), True),
    StructField("manager_name", StringType(), True),
    StructField("performance_rating", StringType(), True),
    StructField("attendance_percentage", StringType(), True),
    StructField("active", BooleanType(), True)
])
```

```python
df = spark.read.format("csv")\
    .option("header",True)\
    .option("inferSchema",False)\
    .schema(myschema) \
    .option("mode","PERMISSIVE")\
    .load("/Volumes/dbacademy/default/myvolume/employee_management_20000.csv")

display(df.limit(5))
```

# STEP 11: Handling Negative Values

```python
df = df.withColumn("experience",when(col("experience") < 0,lit(0))
                                .otherwise(col("experience")))
```

## STEP 15.2: Wide Transforamtion


### 1: Group by and Order By


```python
data1 = df.groupBy("department").agg(count("emp_id").alias("Total_Employees"),sum("salary").alias("Total_Salary")) \
          .orderBy(col("Total_Employees").desc(),col("Total_Salary").desc())

display(data1)
```

```python
data2 = df.filter(col("state") == "Rajasthan")

data2.groupBy("designation").agg(round(mean("salary"),2).alias("Average_Salary")).show()
```

```python
data3 = df.groupBy("city","designation").agg(count("emp_id").alias("Total_Employees"),max("salary").alias("Max_Salary"),
                                        min("salary").alias("Min_Salary")) \
                                        .orderBy(col("city"),col("designation"))

display(data3)
```

# STEP 17: Complete Pipeline

```python
pipeline_df = (
    # READ
    spark.read.format("csv")\
    .option("header",True)\
    .option("mode","PERMISSIVE")\
    .schema(myschema) \
    .load("/Volumes/dbacademy/default/myvolume/employee_management_20000.csv")

    # TRANSFORM
    .withColumn("age",col("age").cast("int"))
    .withColumn("salary",col("salary").cast("int"))
    .withColumn("experience",col("experience").cast("int"))
    .withColumn("performance_rating", col("performance_rating").cast("int"))
    .withColumn("attendance_percentage", col("attendance_percentage").cast("double"))

    # Remove records having null employee names
    .dropna(subset = ['employee_name'])

    #  Fill remaining null values
    .fillna({
    "manager_name": "Not Assigned",
    "performance_rating": 0,
    "attendance_percentage":0,
    "active": False})

    # Handle invalid values
    .withColumn("experience",when(col("experience") < 0,lit(0))
                                .otherwise(col("experience")))
    .withColumn("salary",when(col("salary") < 0, col("salary")*(-1))
                            .otherwise(col("salary")))

    .withColumn("age",when(col("age")<0,avg_age)
                        .otherwise(col("age")))
    
    # Remove duplicates
    .dropDuplicates()

    # Rename columns
    .withColumnsRenamed({"employee_id":"emp_id",
                       "employee_name":"emp_name"})
    
    # Create new columns
    .withColumn("Experience_Level",when(col("experience") <= 1,"Junior")
                                      .when(col("experience") <= 4,"Mid Level")
                                      .otherwise("Senior"))
    .withColumn("attendance_status",when(col("attendance_percentage") >= 90, "Excellent")
                                      .otherwise("Needs Improvement"))
    
     # Filter                         
     .filter(
        (col("attendance_percentage") > 90) & (col("designation") == "Engineer"))
    
    # Select required columns
    .select(
        "emp_id","emp_name","department","designation","salary","attendance_percentage","Experience_Level"
        ,"attendance_status","city","state")
)
```

**Insights**

Executed an end-to-end data pipeline including read, transform, filter, and write.

```python
 # Write CSV File
pipeline_df.write.format("csv")\
            .mode("overwrite")\
            .option("header","true")\
            .save("/Volumes/dbacademy/default/myvolume/pipeline_csv_output")
```

```python
# Write Parquet File
pipeline_df.write.format("parquet")\
    .mode("overwrite")\
    .option("header","true")\
    .save("/Volumes/dbacademy/default/myvolume/pipeline_parquet_output")
```

**Insights**

Stored the final processed data as a CSV file and parquet file.
