## STEP 7: Data Cleaning(Handling Null Values)

### STEP 7(a): Drop rows

```python
cleaned_df = df.na.drop()
cleaned_df.show()
```

```python
cleaned_df.sort(col('customer_id')).show()
```

### STEP 7(b): Fill null values

```python
df.select(
    [count(when(col(x).isNull(), x)).alias(x) for x in df.columns]
).show()
```

```python
df = df.fillna({
"customer_name":"Unknown",
"age":0,
"gender":"Unknown",
"city":"Unknown",
"region":"Unknown",
"membership":"Unknown",
"total_orders":0,
"total_spent":0,
"status":"Unknown"
})
```

### STEP 7(c): Replace Empty Strings

```python
df = df.withColumn("customer_name",when(col("customer_name")=="","Unknown").otherwise(col("customer_name")))
df = df.withColumn("gender",when(col("gender")=="","Unknown").otherwise(col("gender")))
df = df.withColumn("city",when(col("city")=="","Unknown").otherwise(col("city")))
df = df.withColumn("region",when(col("region")=="","Unknown").otherwise(col("region")))
df = df.withColumn("membership",when(col("membership")=="","Unknown").otherwise(col("membership")))
df = df.withColumn("status",when(col("status")=="","Unknown").otherwise(col("status")))
```

### STEP 7(d): Filtering

```python
df = df.filter(col('age') > 25)
df = df.filter((col('city') != 'Unknown') & (col('region') != 'Central'))
df = df.filter(col('total_spent') > 0)
```

### STEP 7(e): Rename Columns

```python
df = df.withColumnRenamed('total_orders','customer_orders')
df = df.withColumnRenamed('total_spent','expenditure')
```

```python
df.show()
```

**Insights:** 

Managed null values by eliminating rows with missing data using dropna() and filling remaining null entries with suitable values through fillna(), and also renamed the column names to more meaningful and descriptive labels using withColumnRenamed(). This improves dataset readability.

## STEP 9: GROUP BY

```python
print("Region-wise Average Expenditure")
df.groupBy("region")\
  .agg(avg("expenditure").alias("avg_expenditure")).show()
```

```python
print("City wise total customer orders")
df.groupBy("city")\
  .agg(sum('customer_orders').alias('total_orders')).show()
```

```python
df.groupBy("membership")\
  .agg(count("*").alias("total_count"),sum("expenditure").alias("total_spent"))\
   .filter(col("total_spent") > 15000)\
  .show()
 
```

**Insights:** 

Used groupBy() with aggregation functions to perform region-wise average expenditure analysis, city-wise total customer order calculation, and membership-wise customer count and total expenditure.

## STEP 10: TRANSFORAMTIONS

#### NARROW TRANSFORMATION - filter(),select(),withColumn(),drop(),dropDuplicates()

```python
df.filter(col("expenditure") > 80000).show()
```

```python
df.withColumn("status", when(col("expenditure") > 80000, "High Value")\
                   .when(col("expenditure") > 50000, "Medium Value")\
                   .otherwise("Low Value")).show()
```

**Insights:** 

Used filter() to retrieve high-spending customers and withColumn() with when() to classify customers into different value categories. These are narrow transformations as they operate on each row independently without redistributing data across partitions.

#### WIDE TRANSFORMATION - groupBy(),join(),orderBy(),distinct(),repartition()

```python
df.groupBy("region","gender")\
  .agg(sum("expenditure").alias("total_spent"))\
  .orderBy(col("total_spent").desc())\
  .show()
```

```python
df.orderBy(col("customer_orders").desc()).show()
```

**Insight:** 

Applied groupBy() and orderBy() as wide transformations, where Spark redistributes data between partitions during execution. This allows efficient aggregation and sorting of records for better analysis.

## STEP 11: COMPLETE PIPELINE

```python
pipeline = (
df
.dropDuplicates() 

.fillna({
"customer_name":"Unknown",
"age":0,
"gender":"Unknown",
"city":"Unknown",
"region":"Unknown",
"membership":"Unknown",
"customer_orders":0,
"expenditure":0,
"status":"Unknown"
})

.filter(col('age') > 25)
.filter((col('city') != 'Unknown') & (col('region') != 'Central'))
.filter(col('expenditure') > 0)

.groupBy("membership")
    .agg(
        count("*").alias("total_customers"),
        sum("expenditure").alias("total_spent"),
        avg("expenditure").alias("avg_spent")
    )

    .filter(col("total_spent") > 15000)
)

pipeline.show()
```

**Insights:**

This pipeline combines all the major data processing steps into a single workflow. It starts by removing duplicate records and handling missing values, then filters the required customer data based on specific conditions. Finally, the data is grouped by membership type to calculate the total number of customers, total expenditure, and average expenditure. This demonstrates how Spark can efficiently clean, transform, and summarize large datasets in a single pipeline.
