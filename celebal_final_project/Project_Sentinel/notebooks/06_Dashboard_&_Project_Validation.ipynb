# Notebook 6 : Dashboard & Project Validation

### Objective

##### This notebook validates the complete Medallion Architecture pipeline by verifying all layers and presenting business KPIs using the Gold and Fraud tables. It serves as the final reporting and validation notebook for the project.

## Step 1 : Import Libraries

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

## Step 2 - Configuration

```python
gold_daily_table = "gold_daily_summary"
gold_bank_table = "gold_bank_summary"
gold_city_table = "gold_city_summary"
gold_transaction_table = "gold_transaction_type_summary"
fraud_table = "gold_fraud_transactions"
```

## Step 3 : Read Tables

```python
daily_summary = spark.table(gold_daily_table)
bank_summary = spark.table(gold_bank_table)
city_summary = spark.table(gold_city_table)
transaction_type_summary = spark.table(gold_transaction_table)
fraud_table = spark.table(fraud_table)
```

## Step 4 : Pipeline Validation

### 4.1 Check Available Tables

```python
display(spark.sql("show tables"))
```

### 4.2 Record Count Validation

```python
print("Bronze Records :",spark.table("bronze_upi_transactions").count())
print("Silver Records :",spark.table("silver_upi_transactions").count())
print("Fraud Records :",fraud_table.count())
```

## Step 5 : Dashboard Queries

```python
print("Daily Summary Records: ")
display(daily_summary)
```

```python
print("Bank Summary Records: ")
display(bank_summary)
```

```python
print("City Summary Records: ")
display(city_summary)
```

```python
print("Transaction Summary Records: ")
display(transaction_type_summary)
```

## Step 6 : Dashboard KPIs

### 6.1 Total Transactions

```python
daily_summary.select(sum("total_transactions")).show()
```

## 6.2 Total Transaction Amount

```python
daily_summary.select(sum("total_amount")).show()
```

## 6.3 Average Transaction Amount

```python
daily_summary.select(round(avg("average_transaction_amount"),2)).show()
```

### 6.4 Overall Success Rate

```python
daily_summary.select(round(avg("success_rate"),2)).show()
```

### 6.5 Overall Failure Rate

```python
daily_summary.select(round(avg("failure_rate"),2)).show()
```

## Step 7 : Fraud Dashboard

```python
display(fraud_table.groupBy("fraud_reason")
        .count()
        .orderBy(col("count").desc()))
```
