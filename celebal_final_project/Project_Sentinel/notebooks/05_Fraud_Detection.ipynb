# Notebook 5 - Fraud Detection

#### Objective

###### This notebook identifies suspicious UPI transactions using rule-based fraud detection techniques.The detected records are stored separately for further investigation.

## Step 1 : Import Libraries

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

## Step 2 : Configuration

```python
silver_table = "silver_upi_transactions"

fraud_table = "gold_fraud_transactions"
```

## Step 3 : Read Silver Table

```python
silver_df = spark.table(silver_table)

display(silver_df.limit(10))
```

## Fraud Rule 1 : High Value Transactions

```python
print("count of High Value Transactions are: ")
silver_df.filter(col("amount") > 48000).count()
```

```python
high_value_fraud = (silver_df.filter(col("amount") > 48000)
    .withColumn("fraud_reason", lit("High Value Transaction")))
```

```python
display(high_value_fraud.limit(5))
```

## Fraud Rule 2 : High Response Time

```python
print("count of High Response Time are: ")
silver_df.filter(col("response_time_ms") > 2850).count()
```

```python
high_time_fraud = (silver_df.filter(col("response_time_ms") > 2850)
                  .withColumn("fraud_reason",lit("High Response Time")))
```

```python
display(high_time_fraud.limit(5))
```

## Fraud Rule 3 : Multiple Failed Transactions

### 3.1 : Find Failed Senders

```python
3failed_sender = (silver_df.filter(col("transaction_status") == "FAILED")
                .groupBy("sender_upi")
                .agg(count("*").alias("failed_transaction_count")))
```

## 3.2 : Only Suspicious Senders

```python
suspicious_failed_senders = (failed_sender.filter(col("failed_transaction_count") >= 2))
```

```python
print("Number of suspicious_failed_senders are:")
display(suspicious_failed_senders.count())
```

## 3.3 – Get Fraud Transactions

```python
failed_transaction = (silver_df.join(suspicious_failed_senders,
                      on="sender_upi",how="inner")
                      .withColumn("fraud_reason",lit("Multiple Failed Transactions"))
                      .drop("failed_transaction_count"))
```

```python
display(failed_transaction)
```

## Step 4 : Combine Fraud Records

### 4.1 Combine All Rules

```python
fraud_df = (high_value_fraud
    .unionByName(high_time_fraud)
    .unionByName(failed_transaction))
```

### 4.2 Remove Duplicate Transactions

```python
fraud_df = fraud_df.dropDuplicates(["transaction_id"])
```

### 4.3 Verify

```python
display(fraud_df.limit(10))
```

```python
print("Number of Records are: ", fraud_df.count())
```

## Step 5 : Write Fraud Table

```python
fraud_df.write \
             .format("delta") \
             .mode("overwrite") \
             .saveAsTable(fraud_table)

print("Fraud Table Created Successfully")
```

## Step 6 : Verify Fraud Table

```python
f_table = spark.table(fraud_table)
display(f_table.limit(5))
```

```python
print("Fraud Records :", f_table.count())
```
