## Import Required Libraries

```python
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()
```

# Part 1: Data Generation

## 1. Customer DataFrame

```python
customer_records = 500

customer_types = ["REGULAR","PREMIUM","VIP"]
customers = []

for i in range(1, customer_records + 1):
    if random.random() < 0.02:
        em = random.choice([
            fake.user_name() + "gmail.com",
            fake.user_name() + "@"
        ])
    else:
        em = fake.email()

    customers.append({
        "customer_id": i,
        "customer_name": f"{fake.first_name()} {fake.last_name()}",
        "email": em,
        "registration_date": fake.date_between("-3y", "today"),
        "customer_type": random.choices(customer_types, weights=[70, 20, 10], k=1)[0]
    })

customer_df = pd.DataFrame(customers)

customer_df.to_csv("customers.csv",index=False)
print("Customer DataFrame Successfully Generated")
```

## 2. Product Dataframe
