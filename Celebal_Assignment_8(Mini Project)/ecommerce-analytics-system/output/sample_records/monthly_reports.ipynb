## Import Required Libraries

```python
import random
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()
```

## Database Creation

```python
import sqlite3
conn = sqlite3.connect("ecommerce.db")
print("Successfully Database Connected")
cursor = conn.cursor()
```

## Helper Function

```python
def run_query(query):
   return pd.read_sql_query(query,conn)
```

## Table Creation

```python
customers = pd.read_csv("customers.csv")
products = pd.read_csv("cleaned_products.csv")
orders = pd.read_csv("cleaned_orders.csv")
order_items = pd.read_csv("order_itmes.csv")
```

```python
customers.to_sql("customers",conn,if_exists="replace",index=False)

products.to_sql("products",conn,if_exists="replace",index=False)

orders.to_sql("orders",conn,if_exists="replace",index=False)

order_items.to_sql("order_items",conn,if_exists="replace",index=False)

print("All tables loaded successfully.")
```

## 1. User Input

```python
report_type = input("Enter Report Type (Daily/Weekly/Monthly): ")

start_date = input("Enter Start Date (YYYY-MM-DD): ")

end_date = input("Enter End Date (YYYY-MM-DD): ")
```

## 2. Total Orders

```python
query = f"""
select count(*)
from orders
where date(order_date)
between '{start_date}' and '{end_date}'
"""

df_query = run_query(query)
df_query
```

## 3. Total Revenue

```python
query = """SELECT COUNT(*)
FROM order_items oi
JOIN orders o
ON oi.order_id = o.order_id
WHERE DATE(o.order_date)
BETWEEN '2025-01-01' AND '2025-01-31';
"""

df_query = run_query(query)
df_query
```

## 4. Unique Customers

```python
query = f"""select count(distinct customer_id) as unique_customers
from orders
where date(order_date)
between '{start_date}'
and '{end_date}'
"""

df_query = run_query(query)
df_query
```

## 5. Top 3 Products

```python
query = f"""select p.product_name, sum(oi.quantity) quantity
from products p
join order_items oi
on p.product_id=oi.product_id
join orders o
on oi.order_id=o.order_id
where date(order_date)
between '{start_date}' and '{end_date}'
group by p.product_name
order by quantity DESC
limit 3
"""
df_query = run_query(query)
df_query
```
